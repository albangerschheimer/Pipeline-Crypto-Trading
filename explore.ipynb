```python
from  datetime import datetime, timezone

def parse_klines(klines):
    parsed_klines = []
    for kline in klines:
        parsed_kline = {
            "open_time": datetime.fromtimestamp(kline[0] / 1000, tz=timezone.utc),
            "open": float(kline[1]),
            "high": float(kline[2]),
            "low": float(kline[3]),
            "close": float(kline[4]),
            "volume": float(kline[5]),
            "close_time": datetime.fromtimestamp(kline[6] / 1000, tz=timezone.utc),
            "quote_asset_volume": float(kline[7]),
            "number_of_trades": int(kline[8]),
            "taker_buy_base_asset_volume": float(kline[9]),
            "taker_buy_quote_asset_volume": float(kline[10]),
        }
        parsed_klines.append(parsed_kline)
    return parsed_klines

parse_klines(klines)
```

```python
def get_klines(symbol, interval="1m", limit=500):
    params = {"symbol": symbol, "interval": interval, "limit": limit}
    response = httpx.get("https://api.binance.com/api/v3/klines", params=params)
    response.raise_for_status()
    return parse_klines(response.json())

get_klines("ETHUSDT", limit=3)
```

```python
params = {"vs_currency": "usd", "order": "market_cap_desc", "per_page": 50, "page": 1}
response = httpx.get("https://api.coingecko.com/api/v3/coins/markets", headers=headers, params=params)
response.raise_for_status()
top50 = response.json()

binance_usdt_symbols = {
    s["symbol"] for s in symbols
    if s["quoteAsset"] == "USDT" and s["status"] == "TRADING"
}

univers = []
for coin in top50:
    symbole_hypothetique = coin["symbol"].upper() + "USDT"
    if symbole_hypothetique in binance_usdt_symbols:
        univers.append({
            "coingecko_id": coin["id"],
            "symbol": coin["symbol"],
            "binance_symbol": symbole_hypothetique,
            "market_cap_rank": coin["market_cap_rank"],
        })

print(len(univers))
for u in univers:
    print(u)
```
